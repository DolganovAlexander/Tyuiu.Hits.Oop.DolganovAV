<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс CreditCard в C#, который будет представлять информацию
о кредитных картах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма

Требования к базовому классу CreditCard:
• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок
действия (ExpiryDate).
• Методы:
o GetInfo(): метод для получения информации о кредитной карте в виде
строки.
o Pay(): метод для оплаты покупки с использованием карты.
o CheckBalance(): метод для проверки баланса на карте.
Требования к производным классам:
1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты,
такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть
переопределен для добавления информации о получении бонусных миль
при оплате покупки.
2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные
атрибуты, такие как Ассистент поддержки (SupportAssistant).
Метод CheckBalance() должен быть переопределен для предоставления
возможности обратиться за помощью к ассистенту поддержки в случае
проблем с балансом.
3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Компания (Company).
Метод GetInfo() должен быть переопределен для включения информации о
компании в описании карты.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class CreditCard
{
    public string CardNumber;
    public string HolderName;
    public DateTime ExpiryDate;

    public virtual string GetInfo()
    {
        return $"Номер карты: {CardNumber}, Холдер: {HolderName}, Срок действия: {ExpiryDate:MM.yyyy}";
    }
    
    public virtual void Pay()
    {
        Console.WriteLine($"Оплата покупки с карты {CardNumber}");
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Проерка баланса карты {CardNumber}");
    }
}

public class GoldenCard : CreditCard
{
    public int BonusMiles;
    public override void Pay()
    {
        Console.WriteLine($"Покупка с Gold-карты {CardNumber}");
        BonusMiles += 200;
        Console.WriteLine($"За покупку начислено 200 бонусных миль.");
        Console.WriteLine($"Всего бонусных миль: {BonusMiles}");
    }
}

public class PremiumCard : CreditCard
{
    public string SupportAssistant;
    public override void CheckBalance()
    {
        Console.WriteLine($"Проверка баланса Premium-карты {CardNumber}");
        Console.WriteLine($"При вознекновении проблем можете обратиться к ассистенту поддержки - {SupportAssistant}.");
    }
}

public class CorporateCard : CreditCard
{
    public string Company;
    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}";
    }
}

    
GoldenCard goldCard = new GoldenCard();
goldCard.CardNumber = "1234-5678-9012-3456";
goldCard.HolderName = "Марк Серков";
goldCard.ExpiryDate = new DateTime(2029, 12, 1);
goldCard.BonusMiles = 200;
Console.WriteLine(goldCard.GetInfo());
goldCard.Pay();

Console.WriteLine();
PremiumCard premiumCard = new PremiumCard();
premiumCard.CardNumber = "1111-3333-5555-8888";
premiumCard.HolderName = "Андрей Зелинский";
premiumCard.ExpiryDate = new DateTime(2029, 12, 1);
premiumCard.SupportAssistant = "Анна";

Console.WriteLine(premiumCard.GetInfo());
premiumCard.CheckBalance();

Console.WriteLine();
CorporateCard corporateCard = new CorporateCard();
corporateCard.CardNumber = "9999-9999-9999-9998";
corporateCard.HolderName = "Дмитрий Ицков";
corporateCard.ExpiryDate = new DateTime(2033, 6, 1);
corporateCard.Company = "ООО RRepiv";

Console.WriteLine(corporateCard.GetInfo());
corporateCard.Pay();
```
